### Assignment 3: Data Cleaning

**Q1.** Introduce NaN values and fill them with the column mean.

```python
df = pd.DataFrame(np.random.randint(1, 50, size=(5, 3)), columns=['A', 'B', 'C']).astype(float)
df.iloc[0, 1] = np.nan
df.iloc[2, 0] = np.nan
df.iloc[3, 2] = np.nan
print("With NaN:")
print(df)

df_filled = df.fillna(df.mean())
print("\nAfter filling with column mean:")
print(df_filled)
```

**Q2.** 4 columns, 6 rows, introduce NaN, drop rows with any NaN.

```python
df = pd.DataFrame(np.random.randint(1, 50, size=(6, 4)), columns=['A', 'B', 'C', 'D']).astype(float)
df.iloc[1, 2] = np.nan
df.iloc[4, 0] = np.nan
print("With NaN:")
print(df)

df_dropped = df.dropna()
print("\nAfter dropping rows with any NaN:")
print(df_dropped)
```

### Assignment 4: Data Aggregation

**Q1.** Group by 'Category', compute sum and mean of 'Value'.

```python
df = pd.DataFrame({
    'Category': np.random.choice(['A', 'B', 'C'], size=10),
    'Value': np.random.randint(1, 100, size=10)
})
print(df)
print("\nSum and Mean per Category:")
print(df.groupby('Category')['Value'].agg(['sum', 'mean']))
```

**Q2.** Columns 'Product', 'Category', 'Sales'. Total sales per category.

```python
df = pd.DataFrame({
    'Product': [f'P{i}' for i in range(1, 9)],
    'Category': np.random.choice(['Electronics', 'Clothing', 'Grocery'], size=8),
    'Sales': np.random.randint(100, 1000, size=8)
})
print(df)
print("\nTotal sales per Category:")
print(df.groupby('Category')['Sales'].sum())
```

### Assignment 5: Merging DataFrames

**Q1.** Two DataFrames with a common column, merge on it.

```python
left = pd.DataFrame({'ID': [1, 2, 3, 4], 'Name': ['Ali', 'Sara', 'Ahmed', 'Zara']})
right = pd.DataFrame({'ID': [1, 2, 3, 5], 'Marks': [85, 90, 78, 66]})

merged = pd.merge(left, right, on='ID')  # inner join (default)
print("Inner merge:")
print(merged)

print("\nOuter merge:")
print(pd.merge(left, right, on='ID', how='outer'))
```

**Q2.** Two DataFrames with different columns. Concatenate along rows and columns.

```python
d1 = pd.DataFrame({'A': [1, 2, 3], 'B': [4, 5, 6]})
d2 = pd.DataFrame({'C': [7, 8, 9], 'D': [10, 11, 12]})

print("Concat along rows (axis=0):")
print(pd.concat([d1, d2], axis=0))

print("\nConcat along columns (axis=1):")
print(pd.concat([d1, d2], axis=1))
```

### Assignment 6: Time Series Analysis

**Q1.** Datetime index, resample to monthly mean.

```python
idx = pd.date_range('2021-01-01', periods=90, freq='D')
ts = pd.DataFrame({'Value': np.random.randint(1, 100, size=90)}, index=idx)
print(ts.head())
print("\nMonthly mean:")
print(ts.resample('ME').mean())   # agar purana pandas ho to 'M' use karein
```

**Q2.** Index '2021-01-01' to '2021-12-31', rolling mean with window of 7 days.

```python
idx = pd.date_range('2021-01-01', '2021-12-31', freq='D')
ts = pd.DataFrame({'Value': np.random.randint(1, 100, size=len(idx))}, index=idx)
ts['Rolling_Mean_7'] = ts['Value'].rolling(window=7).mean()
print(ts.head(10))
```

### Assignment 7: MultiIndex DataFrame

**Q1.** Create a MultiIndex DataFrame, do indexing and slicing.

```python
arrays = [['A', 'A', 'B', 'B', 'C', 'C'], [1, 2, 1, 2, 1, 2]]
index = pd.MultiIndex.from_arrays(arrays, names=['Letter', 'Number'])
df = pd.DataFrame({'Value': np.random.randint(1, 100, size=6)}, index=index)
print(df)

print("\nAll rows of 'A':")
print(df.loc['A'])

print("\nSpecific row ('B', 2):")
print(df.loc[('B', 2)])

print("\nCross-section: Number == 1")
print(df.xs(1, level='Number'))
```

**Q2.** MultiIndex 'Category' and 'SubCategory', sum of values for each.

```python
index = pd.MultiIndex.from_product([['Fruit', 'Veg'], ['Sub1', 'Sub2']],
                                   names=['Category', 'SubCategory'])
df = pd.DataFrame({'Value': np.random.randint(1, 100, size=4)}, index=index)
print(df)

print("\nSum per Category:")
print(df.groupby(level='Category').sum())

print("\nSum per Category & SubCategory:")
print(df.groupby(level=['Category', 'SubCategory']).sum())
```

### Assignment 8: Pivot Tables

**Q1.** Pivot table: sum of 'Value' for each 'Category' by 'Date'.

```python
df = pd.DataFrame({
    'Date': pd.to_datetime(['2024-01-01', '2024-01-01', '2024-01-02', '2024-01-02', '2024-01-03', '2024-01-03']),
    'Category': ['A', 'B', 'A', 'B', 'A', 'B'],
    'Value': np.random.randint(10, 100, size=6)
})
print(df)
pivot = pd.pivot_table(df, values='Value', index='Date', columns='Category', aggfunc='sum')
print("\nPivot table:")
print(pivot)
```

**Q2.** Pivot table: mean 'Revenue' for each 'Quarter' by 'Year'.

```python
df = pd.DataFrame({
    'Year': [2022]*4 + [2023]*4,
    'Quarter': ['Q1', 'Q2', 'Q3', 'Q4']*2,
    'Revenue': np.random.randint(1000, 5000, size=8)
})
print(df)
pivot = pd.pivot_table(df, values='Revenue', index='Year', columns='Quarter', aggfunc='mean')
print("\nPivot table:")
print(pivot)
```

### Assignment 9: Applying Functions

**Q1.** Apply a function that doubles all values.

```python
df = pd.DataFrame(np.random.randint(1, 10, size=(5, 3)), columns=['A', 'B', 'C'])
print("Original:")
print(df)

def double(x):
    return x * 2

print("\nDoubled:")
print(df.apply(double))
```

**Q2.** Lambda function to create a new column = sum of the existing columns.

```python
df = pd.DataFrame(np.random.randint(1, 10, size=(6, 3)), columns=['A', 'B', 'C'])
df['Sum'] = df.apply(lambda row: row['A'] + row['B'] + row['C'], axis=1)
print(df)
```
