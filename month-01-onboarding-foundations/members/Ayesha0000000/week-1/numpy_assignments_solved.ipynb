### Assignment 2: Array Indexing and Slicing

**1.** Create a NumPy array of shape (6, 6) with values from 1 to 36. Extract the sub-array consisting of the 3rd to 5th rows and 2nd to 4th columns.

```python
arr = np.arange(1, 37).reshape(6, 6)
print("Array:\n", arr)

# 3rd to 5th rows -> indices 2:5, 2nd to 4th columns -> indices 1:4
sub = arr[2:5, 1:4]
print("\nSub-array (rows 3-5, columns 2-4):\n", sub)
```

**2.** Create a NumPy array of shape (5, 5) with random integers. Extract the elements on the border.

```python
arr = np.random.randint(1, 100, size=(5, 5))
print("Array:\n", arr)

# Collect top row, bottom row, then left and right columns without repeating the corners
border = np.concatenate([arr[0, :], arr[-1, :], arr[1:-1, 0], arr[1:-1, -1]])
print("\nBorder elements:", border)

# Same idea with a boolean mask, which also keeps the 2D layout
mask = np.ones(arr.shape, dtype=bool)
mask[1:-1, 1:-1] = False
print("\nBorder only (inside shown as 0):\n", np.where(mask, arr, 0))
```

### Assignment 4: Statistical Operations

**1.** Create a NumPy array of shape (5, 5) filled with random integers. Compute the mean, median, standard deviation, and variance of the array.

```python
arr = np.random.randint(1, 50, size=(5, 5))
print("Array:\n", arr)

print("\nMean              :", np.mean(arr))
print("Median            :", np.median(arr))
print("Standard deviation:", np.std(arr))
print("Variance          :", np.var(arr))
```

**2.** Create a NumPy array of shape (3, 3) with values from 1 to 9. Normalize the array (scale the values to have a mean of 0 and a standard deviation of 1).

```python
arr = np.arange(1, 10).reshape(3, 3)
print("Original array:\n", arr)

# z-score formula: (x - mean) / std
normalized = (arr - arr.mean()) / arr.std()
print("\nNormalized array:\n", np.round(normalized, 4))

# Quick check
print("\nNew mean:", round(normalized.mean(), 10))
print("New std :", round(normalized.std(), 10))
```

### Assignment 5: Broadcasting

**1.** Create a NumPy array of shape (3, 3) filled with random integers. Add a 1D array of shape (3,) to each row of the 2D array using broadcasting.

```python
arr = np.random.randint(1, 10, size=(3, 3))
row_vec = np.array([10, 20, 30])        # shape (3,)
print("2D array:\n", arr)
print("\n1D array:", row_vec)

# (3,3) + (3,) -> the 1D array is stretched across every row
print("\nResult:\n", arr + row_vec)
```

**2.** Create a NumPy array of shape (4, 4) filled with random integers. Subtract a 1D array of shape (4,) from each column of the 2D array using broadcasting.

```python
arr = np.random.randint(1, 20, size=(4, 4))
vec = np.array([1, 2, 3, 4])            # shape (4,)
print("2D array:\n", arr)
print("\n1D array:", vec)

# To subtract from each COLUMN (the vector runs down the rows),
# turn it into a column vector of shape (4, 1) so it stretches across the columns
result = arr - vec.reshape(4, 1)
print("\nResult (vector subtracted from each column):\n", result)
```

### Assignment 6: Linear Algebra

**1.** Create a NumPy array of shape (3, 3) representing a matrix. Compute its determinant, inverse, and eigenvalues.

```python
A = np.array([[4, 2, 1],
              [3, 5, 2],
              [1, 2, 6]])
print("Matrix A:\n", A)

det = np.linalg.det(A)
print("\nDeterminant:", round(det, 4))

# The inverse only exists when the determinant is not zero
if not np.isclose(det, 0):
    inv = np.linalg.inv(A)
    print("\nInverse:\n", np.round(inv, 4))
    print("\nCheck (A x inverse = identity):\n", np.round(A @ inv, 4))

eigenvalues = np.linalg.eigvals(A)
print("\nEigenvalues:", np.round(eigenvalues, 4))
```

**2.** Create two NumPy arrays of shape (2, 3) and (3, 2). Perform matrix multiplication on these arrays.

```python
A = np.array([[1, 2, 3],
              [4, 5, 6]])               # (2, 3)
B = np.array([[7, 8],
              [9, 10],
              [11, 12]])                # (3, 2)

result = np.dot(A, B)                   # same as A @ B, result shape is (2, 2)
print("A:\n", A)
print("\nB:\n", B)
print("\nA x B:\n", result)
print("\nShape of result:", result.shape)
```

### Assignment 9: Structured Arrays

**1.** Create a structured array with fields 'name' (string), 'age' (integer), and 'weight' (float). Add some data and sort the array by age.

```python
dtype = [('name', 'U10'), ('age', 'i4'), ('weight', 'f4')]
people = np.array([('Ayesha', 21, 55.5),
                   ('Maryam', 22, 58.2),
                   ('Ali',    19, 70.1),
                   ('Sara',   25, 62.4),
                   ('Usman',  20, 75.8)], dtype=dtype)
print("Original:\n", people)

sorted_people = np.sort(people, order='age')
print("\nSorted by age:\n", sorted_people)
```

**2.** Create a structured array with fields 'x' and 'y' (both integers). Add some data and compute the Euclidean distance between each pair of points.

```python
points = np.array([(0, 0), (3, 4), (6, 8), (1, 1)],
                  dtype=[('x', 'i4'), ('y', 'i4')])
print("Points:\n", points)

x = points['x']
y = points['y']

# Broadcasting builds the difference between every pair of points at once
dx = x[:, None] - x[None, :]
dy = y[:, None] - y[None, :]
dist = np.sqrt(dx**2 + dy**2)

print("\nDistance matrix (row i, column j = distance between point i and point j):\n", np.round(dist, 3))

# Also list each unique pair
n = len(points)
print("\nPair by pair:")
for i in range(n):
    for j in range(i + 1, n):
        print(f"Point {i} {tuple(points[i])} to Point {j} {tuple(points[j])}: {dist[i, j]:.3f}")
```

### Assignment 10: Masked Arrays

**1.** Create a masked array of shape (4, 4) with random integers and mask the elements greater than 10. Compute the sum of the unmasked elements.

```python
data = np.random.randint(1, 21, size=(4, 4))
masked = ma.masked_greater(data, 10)    # hides everything above 10
print("Original data:\n", data)
print("\nMasked array (-- means hidden):\n", masked)

print("\nSum of unmasked elements:", masked.sum())
```

**2.** Create a masked array of shape (3, 3) with random integers and mask the diagonal elements. Replace the masked elements with the mean of the unmasked elements.

```python
data = np.random.randint(1, 21, size=(3, 3))
mask = np.eye(3, dtype=bool)            # True on the diagonal
masked = ma.masked_array(data, mask=mask)
print("Original data:\n", data)
print("\nMasked array (diagonal hidden):\n", masked)

mean_val = masked.mean()                # mean ignores masked values
print("\nMean of unmasked elements:", round(mean_val, 4))

filled = masked.filled(mean_val)        # put the mean in the masked positions
print("\nAfter replacing masked elements with the mean:\n", np.round(filled, 2))
```
